```python
from Bio import SeqIO
import pandas as pd

def get_fasta_entry_names(fasta_file):
    entry_names = [record.id for record in SeqIO.parse(fasta_file, "fasta")]
    return entry_names
```

```python
fasta_path = "../input/genome/PA14/transcripts.fa"  # Replace with your FASTA file path
names = get_fasta_entry_names(fasta_path)
print(names)
```

```python
len(names)
```

```python
genes = [i.split("-")[1] for i in names]
genes[:5]
```

```python
df = pd.DataFrame(genes)
df['1'] = genes
df.head()
```

```python
df.to_csv("../input/genome/PA14/t2g_simpleaf.txt", sep="\t", index=False, header=False)
```
