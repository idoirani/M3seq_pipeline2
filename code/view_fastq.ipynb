```python
from Bio import SeqIO
import gzip
```

```python
def find_barcode_in_files(barcode, file_paths, case_sensitive=True):
    found = False
    for path in file_paths:
        with open(path, "r") as f:
            for line in f:
                line_barcode = line.strip()
                if not case_sensitive:
                    if barcode.lower() == line_barcode.lower():
                        print(f"Found in: {path}")
                        found = True
                        break
                else:
                    if barcode == line_barcode:
                        print(f"Found in: {path}")
                        found = True
                        break
    if not found:
        print(f"Not found: {barcode}")
```

```python

def print_fastq_head(file_path, num_entries=2, l=8, r=21):
    with gzip.open(file_path, "rt") as handle:  # 'rt' = read text mode
        for i, record in enumerate(SeqIO.parse(handle, "fastq")):
            if i >= num_entries:
                break
            print(f"ID: {record.id}")
            print(f"Sequence: {record.seq}")
            # print(f"RC: {record.reverse_complement().seq}")
            print(len(record.seq))
            print(f'bc:{record.seq[l:r]}')
            find_barcode_in_files(record.seq[l:r], ["../input/platebarcodes/1.txt", "../input/platebarcodes/2.txt", "../input/platebarcodes/3.txt"])
            # print(f"Quality scores: {record.letter_annotations['phred_quality']}")
            # print("-" * 40)
```

```python
def read3_view(file_path, num_entries=2, l=14, r=30):
    with gzip.open(file_path, "rt") as handle:  # 'rt' = read text mode
        for i, record in enumerate(SeqIO.parse(handle, "fastq")):
            if i >= num_entries:
                break
            print(f"ID: {record.id}")
            print(f"Sequence: {record.seq}")
            # print(f"RC: {record.reverse_complement().seq}")
            print(len(record.seq))
            print(f'bc:{record.seq[l:r]}')
            find_barcode_in_files(record.seq[l:r], ["../input/whitelist/rc_737K-cratac-v1.txt", "../input/whitelist/737K-cratac-v1.txt", "../input/whitelist/whitelist_umitools.txt"])
            # print(f"Quality scores: {record.letter_annotations['phred_quality']}")
```
